```python
# Q1. Why are dictionary keys immutable in Python?

# Creating an immutable tuple
tuple_key = (10, 20)

# Creating a dictionary
my_dict = {}

# Using tuple as a dictionary key
my_dict[tuple_key] = "Tuple is allowed as a key"

# Printing the dictionary
print("Dictionary with tuple key:")
print(my_dict)

# Using a mutable type as a key will cause an error
# Uncomment to see the error
# list_key = [10, 20]
# my_dict[list_key] = "List is not allowed as a key"
```

```python
# Q2. Write a program to find the frequency of elements in a list

# Given list
elements = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

# Empty dictionary to store frequency
frequency = {}

# Loop through list elements
for item in elements:
    if item in frequency:
        frequency[item] = frequency[item] + 1
    else:
        frequency[item] = 1

# Printing frequency of each element
print("Frequency of elements:")
print(frequency)
```

```python
# Q4. Merge two dictionaries and keep the maximum value for common keys

# First dictionary
dict1 = {"a": 10, "b": 20}

# Second dictionary
dict2 = {"b": 30, "c": 5}

# Empty dictionary for merged result
merged_dict = {}

# Combine all keys from both dictionaries
all_keys = dict1.keys() | dict2.keys()

# Loop through all keys
for key in all_keys:
    value1 = dict1.get(key, 0)
    value2 = dict2.get(key, 0)

    if value1 > value2:
        merged_dict[key] = value1
    else:
        merged_dict[key] = value2

# Printing merged dictionary
print("Merged dictionary:")
print(merged_dict)
```

```python
# Q5. Create a dictionary of odd numbers and their cubes using dictionary comprehension

# Given list of numbers
numbers = [1, 2, 3, 4, 5, 6, 7]

# Empty dictionary
odd_cube_dict = {}

# Loop through numbers
for num in numbers:
    if num % 2 != 0:
        odd_cube_dict[num] = num ** 3

# Printing the result
print("Odd numbers and their cubes:")
print(odd_cube_dict)
```
